# Project: Predicting Rent Prices in Malaysia
To estimate monthly rental prices in Kuala Lumpur and Selangor, I applied two different models on the same dataset:

- **Linear Regression** assumes a linear relationship between the rent and property features like size, rooms, furnishing, etc.

- **Regression Trees** can model non-linear interactions and capture more complex decision boundaries.

Both models were trained and tested on the same dataset, allowing for a fair, side-by-side comparison of their predictive performance on rent estimation tasks.

Below is the comparison of the two models' performance：

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

data = {
    "Set": ["Train", "Validation", "Test"],
    "Linear Regression - MSE": [146830.82, 148084.08, 138270.30],
    "Linear Regression - MAE": [287.13, 295.19, 282.56],
    "Linear Regression - R2": [0.47, 0.45, 0.47],
    "Regression Tree - MSE": [169224.42, 176618.79, 173802.66],
    "Regression Tree - MAE": [312.42, 318.53, 319.52],
    "Regression Tree - R2": [0.4109, 0.3814, 0.3621]
}

df = pd.DataFrame(data)

def plot_single_metric(metric_name, linear_col, tree_col, ylabel, title):
    x = np.arange(len(df["Set"]))
    width = 0.35

    plt.figure(figsize=(8, 5))
    bars1 = plt.bar(x - width/2, df[linear_col], width, label='Linear Regression')
    bars2 = plt.bar(x + width/2, df[tree_col], width, label='Regression Tree')

    for bar in bars1 + bars2:
        height = bar.get_height()
        plt.annotate(f'{height:.2f}',
                     xy=(bar.get_x() + bar.get_width() / 2, height),
                     xytext=(0, 3),
                     textcoords="offset points",
                     ha='center', va='bottom', fontsize=9)

    plt.ylabel(ylabel)
    plt.title(title)
    plt.xticks(x, df["Set"])
    plt.legend()
    plt.grid(axis='y', linestyle='--', alpha=0.6)
    plt.tight_layout()
    plt.show()

plot_single_metric("MSE", "Linear Regression - MSE", "Regression Tree - MSE", "MSE", "Model Comparison: MSE")
plot_single_metric("MAE", "Linear Regression - MAE", "Regression Tree - MAE", "MAE", "Model Comparison: MAE")
plot_single_metric("R2", "Linear Regression - R2", "Regression Tree - R2", "R² Score", "Model Comparison: R² Score")
```

## MSE Comparison
Mean Squared Error (MSE) is a measure of prediction error magnitude.

Across train, validation, and test sets, Linear Regression consistently has lower MSE than the Regression Tree. This suggests Linear Regression is more stable in minimizing large errors across all data splits.

---

## MAE Comparison
Mean Absolute Error (MAE) represents the average size of errors.

Linear Regression again performs better (lower MAE) than Regression Trees across all sets. This indicates it provides more accurate rent predictions on average, even if the errors are small.

---

## R² Score Comparison
R² Score measures how well the model explains variance in the target variable.

Linear Regression scores higher in R² across the board (0.47) vs. Regression Tree (0.41).This shows that Linear Regression captures more of the underlying trend in rental prices.

---

## Conclusion:
Based on MSE, MAE, and R² across train, validation, and test sets, **Linear Regression outperforms Regression Trees on this dataset**, offering more accurate and reliable predictions of rental prices.

# Project: Predicting Student Depression Risks

To identify students at risk of depression, I applied six different classification models on the same dataset:

- **The Perceptron** is a simple linear classifier that makes predictions based on a weighted sum of the input features.

- **Logistic Regression** models the probability that a student is depressed using a logistic (sigmoid) function. It provides interpretable coefficients and is widely used for binary classification problems.

- **Neural Networks** can model complex non-linear relationships by stacking layers of nodes (neurons) that transform inputs through learned weights and activation functions.

- **K-Nearest Neighbors (KNN)** predicts a student’s label by looking at the labels of the closest K students in the dataset.

- **Random Forests** build an ensemble of decision trees using randomly selected features and samples.

- **Gradient Boosting** sequentially builds trees that correct the errors of previous ones.

All models were trained and tested on the same dataset, allowing for a fair, side-by-side comparison of their classification performance on student depression risk prediction.

Below is the comparison of the six models' performance:

```python
import matplotlib.pyplot as plt
import numpy as np

models = ['Perceptron', 'Logistic\nRegression', 'Neural\nNetworks', 'KNN', 'Random\nForests', 'Gradient\nBoosting']
datasets = ['Train', 'Validation', 'Test']
x = np.arange(len(models))
width = 0.25

accuracy_scores = {
    'Train':     [0.77, 0.75, 0.59, 0.85, 0.78, 0.59],
    'Validation':[0.77, 0.75, 0.58, 0.79, 0.77, 0.58],
    'Test':      [0.76, 0.77, 0.59, 0.80, 0.78, 0.59]
}

def plot_metric(metric_dict, metric_name):
    fig, ax = plt.subplots(figsize=(12, 6))
    for i, split in enumerate(datasets):
        y = metric_dict[split]
        bars = ax.bar(x + i * width - width, y, width, label=split)
        for j, val in enumerate(y):
            ax.text(j + i * width - width, val + 0.005, f'{val:.2f}', ha='center', va='bottom', fontsize=8)

    ax.set_ylabel(metric_name)
    ax.set_title(f'Model Comparison: {metric_name}')
    ax.set_xticks(x)
    ax.set_xticklabels(models)
    ax.legend()
    plt.ylim(0, 1.05)
    plt.grid(True, axis='y', linestyle='--', alpha=0.6)
    plt.tight_layout()
    plt.show()

plot_metric(accuracy_scores, 'Accuracy')
```

Accuracy Summary:

KNN achieved the highest accuracy (Train: 0.85, Test: 0.80), showing strong and consistent performance.

Random Forests followed closely (around 0.78), with good generalization.

Perceptron and Logistic Regression showed moderate and stable performance (around 0.75–0.77).

Neural Networks and Gradient Boosting had the lowest accuracy (around 0.58–0.59).
